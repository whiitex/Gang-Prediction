```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch_geometric.nn import GCNConv, SAGEConv, GATv2Conv
import torch.optim as optim

import os
import sys
import numpy as np
import matplotlib.pyplot as plt
from torch_geometric.datasets import Planetoid

sys.path.append(os.path.abspath(os.path.join(os.getcwd(), "src")))

from graph_utils import *
from coarsening_utils import *
from utils.utils import *
from utils.visualization import *
from utils.aml_data_retriever import load_aml_dataset
from coarsening_aware_loss import *

import warnings
warnings.filterwarnings("ignore")

# torch geometric
from torch_geometric.data import Data

from sklearn.metrics import accuracy_score, classification_report
```

```python
class GCN(nn.Module):
    def __init__(self, nfeat, nhid, nclass, dropout=0.5):
        super(GCN, self).__init__()
        self.conv1 = SAGEConv(nfeat, nhid)
        self.conv2 = SAGEConv(nhid, nclass)
        self.dropout = dropout

    def forward(self, x, edge_index):
        x = F.relu(self.conv1(x, edge_index))
        x = F.dropout(x, self.dropout, training=self.training)
        x = self.conv2(x, edge_index)
        return F.log_softmax(x, dim=1)

    def get_embeddings(self, x, edge_index):
        x = F.relu(self.conv1(x, edge_index))
        return x

    def reset_parameters(self):
        self.conv1.reset_parameters()
        self.conv2.reset_parameters()


def train_gnn_1_epoch(
        model: nn.Module,
        optimizer: optim.Optimizer,
        criterion: nn.Module,
        data: Data,
        train_idx: list,
        val_idx: list
):
    
    # train
    model.train()
    optimizer.zero_grad()

    output = model(data.x, data.edge_index)

    loss = criterion(
        output, data.embeddings, data.y, train_idx, coarse_loss=False
    )
    loss.backward()
    optimizer.step()
    train_acc = accuracy_score(
        data.y[train_idx].detach().cpu().numpy(),
        output[train_idx].max(1)[1].detach().cpu().numpy(),
    )

    # validate
    model.eval()
    with torch.no_grad():
        output = model(data.x, data.edge_index)

        loss_val = criterion(
            output, data.embeddings, data.y, val_idx, coarse_loss=False
        )
        acc_val = accuracy_score(
            data.y[val_idx].cpu().numpy(),
            output[val_idx].max(1)[1].cpu().numpy()
        )

    return loss.item(), train_acc, loss_val.item(), acc_val
```

```python
def train_GNN_coarsening_aware_loss_vanilla(
    data: Data,
    lr=0.01,
    wd=5e-4,
    method="variation_neighborhoods",
    nhid=128,
    dropout=0.1,
):
    # general parameters
    device = data.x.device
    N = data.num_nodes
    nclass = len(torch.unique(data.y))
    nfeat = data.x.shape[1]

    # model
    model = GCN(
        nfeat=nfeat,  
        nhid=nhid,
        nclass=nclass,
        dropout=dropout
    ).to(device)

    # data split
    train_idx, val_idx, test_idx = create_train_val_test_split(N)
    train_idx = train_idx.to(device)
    val_idx = val_idx.to(device)
    test_idx = test_idx.to(device)

    # criterion
    criterion = CoarseningAwareLoss()

    # optimizer
    optimizer = optim.Adam(model.parameters(), lr=lr, weight_decay=wd)    

    # train and validate the model
    yacc_t, yloss_t, yloss_v, yacc_v = [], [], [], []
    
    for epoch in range(1, 101):
        train_loss, train_acc, val_loss, val_acc = train_gnn_1_epoch(
            model,
            optimizer,
            criterion,
            data,
            train_idx,
            val_idx
        )

        if epoch % 10 == 0:
            print(
                f"Epoch: {epoch:03d}, "
                f"Train Loss: {train_loss:.4f}, "
                f"Train Acc: {train_acc:.4f}, "
                f"Val Loss: {val_loss:.4f}, "
                f"Val Acc: {val_acc:.4f}"
            )

        yacc_t.append(train_acc)
        yloss_t.append(train_loss)
        yloss_v.append(val_loss)
        yacc_v.append(val_acc)


    name = f"data_vanilla_lr_{lr:.2e}_wd_{wd:.2e}_nhid_{nhid}.npy"
    np.save(
        f"./results/vanilla/{name}",
        {
            "yacc_t": yacc_t,
            "yloss_t": yloss_t,
            "yloss_v": yloss_v,
            "yacc_v": yacc_v,
        },
    )
```

```python
def create_pyg_data(features, edges_idx, edge_features, labels, weights=None) -> Data:
    # edge_index, edge_attr = to_undirected(edges_idx, edge_features)
    if weights is None:
        weights = torch.ones(edges_idx.shape[1], device=labels.device)

    return Data(
        x=features.float(),
        edge_index=edges_idx.long(),
        edge_attr=edge_features.float(),
        y=labels.long().view(-1),
        num_nodes=features.shape[0],
        edge_weight=weights.float()
    )
```
